```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)

print("OpenAI client ready:", api_key is not None)
```

```python
AGENT_INSTRUCTIONS = """
You are an Unreal Engine Technical Assistant.

ROLE:
You are a technical assistant specializing in Unreal Engine.

GOAL:
Help users diagnose Unreal Engine technical-art problems.

RULES:
- Explain the problem clearly.
- Identify likely causes.
- Give practical troubleshooting steps.
- Use Unreal Engine terminology when appropriate.
- Do not invent settings or features.
- If information is missing, clearly say what information is required.

OUTPUT FORMAT:

Problem:
Likely Cause:
Steps:
Additional Information Needed:
"""
```

```python
user_input = input("You :")
```

```python
response = client.responses.create(
    model="gpt-5.4-mini",
    instructions=AGENT_INSTRUCTIONS,
    input=user_input
)

print(response.output_text)
```
